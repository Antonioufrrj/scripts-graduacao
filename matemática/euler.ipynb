```python
import numpy as np
import matplotlib.pyplot as plt
```

\begin{align}
EDO \ DE \ PRIMEIRA \ ORDEM:\\ \frac{dy}{dx} = \frac{1}{(1 + x^2)} - 2y \,\, condição \ inicial  \ y(0) = 0 \\
solução \ analitica \, \, y = \frac{x}{1 + x^2}
\end{align}

```python
def dydx(x, y) : # funcao que recebe x e y e retorna o valor da edo no respectivos pontos
    return (1/(1 + x*x) - 2*y*y);
```

```python
def euler(t0, y0, t, n):
    x = np.zeros(n)
    y = np.zeros(n)
    h = (t - t0) / n
    y[0] = y0
    #print(x[0], y[0])
    for i in range(1,n):
        y[i] = y[i-1] + dydx(x[i-1],y[i-1]) * h
        x[i] = x[i-1] + h
        #print(x[i], y[i])
    return x,y
```

```python
t0 = 0.0          # t inicial
y0 = 0            # condicao inicial
t  = 2
n  = 50
solution = euler(t0, y0, t, n)
```

```python
solution
```

```python
solution[0]
# analitica e o vetor responsavel por armazenar a solucao analitica da edo em cada ponto
analitica =  solution[0]/(1 + solution[0]*solution[0])
analitica
```

```python
# plotando o grafico para comprar o resultado do metodo com a solução analitica
plt.figure( figsize=(9, 9))
plt.plot(solution[0], solution[1], color='red', label='euler')
plt.plot(solution[0], analitica, color='yellow', label='analitica')
plt.title("analise preçisão")
plt.legend()
plt.axis([0, 2.0, 0.00, 0.55])
plt.show()
```

```python
plt.plot(solution[0], abs(analitica - solution[1]), color='red', label='euler')
plt.axis([0, 2, 0, 0.03])
plt.show()
```
